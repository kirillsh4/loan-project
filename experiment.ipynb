```python
import pandas as pd
import numpy as np

df = pd.read_csv(f"{path}/loan_data.csv")
```

```python
df.head()
df.describe()
```

```python

def inject_missing_values(df, col_missing_map, random_state=42):
    """
    df: pandas DataFrame
    col_missing_map: dict like {'col1': 0.05, 'col2': 0.1}
    """
    df_new = df.copy()
    rng = np.random.default_rng(random_state)

    for col, frac in col_missing_map.items():
        n_missing = int(len(df_new) * frac)
        missing_idx = rng.choice(df_new.index, size=n_missing, replace=False)
        df_new.loc[missing_idx, col] = np.nan

    return df_new

column_missing_map = {
    # leave these complete / almost complete
    "person_age": 0.00,
    "person_gender": 0.004,
    "person_education": 0.011,
    "person_income": 0.032,
    "person_emp_exp": 0.028,
    "person_home_ownership": 0.011,
    "loan_amnt": 0.00,
    "loan_intent": 0.018,
    "loan_int_rate": 0.034,
    "loan_percent_income": 0.00,   # better treated as derived
    "cb_person_cred_hist_length": 0.018,
    "credit_score": 0.032,
    "previous_loan_defaults_on_file": 0.004,
    "loan_status": 0.00            # keep target complete
}

df_missing = inject_missing_values(df, column_missing_map)
df_missing.to_csv("loan_data_raw.csv", index=False)
```
